## 1 · Bäume als eigener 3D-Layer

Laden Sie 13.317 automatische Baumkandidaten aus neun 1-km²-Kacheln. Nutzen Sie im Viewer den Höhenmodus, um zwischen einheitlichen 15 m und geschätzter Baumhöhe zu wechseln. Schalten Sie den Layer aus und wieder ein; klicken Sie einen Baum an, um seine ID und Höhenangaben zu prüfen.


```python
from basemap3d import Map, TreeLayer, WmsTileLayer
m = Map(location=[49.99892, 8.27412], zoom_start=16,
        title="Mainz · Bäume als 3D-Layer")
WmsTileLayer(opacity=0.7).add_to(m)
TreeLayer(variant="uniform").add_to(m)
m.save("results/mainz_trees.html")
m
```

## 2 · Geschätzte Höhe als Startansicht

Erstellen Sie eine unabhängige zweite Karte. Vergleichen Sie die räumliche Verteilung mit der einheitlichen Darstellung. Die Quelle enthält 29 Höhenwerte über 30 m, die geprüft werden sollen. Ein Kandidat ist kein bestätigter Baum.


```python
m_estimated = Map(location=[49.99892, 8.27412], zoom_start=16,
                  title="Mainz · Geschätzte Baumhöhe")
TreeLayer(variant="estimated").add_to(m_estimated)
m_estimated.save("results/mainz_trees_estimated.html")
m_estimated
```

## Einordnung und Transfer

- Der 3 × 3 km große Bearbeitungsrahmen ist keine Verwaltungsgrenze. Es wurden nur Kandidaten innerhalb der verwendeten Mainz-Grenze aufgenommen.
- DOM1/DGM1 (2023) und Luftbild (2024) stammen aus unterschiedlichen Zeitpunkten. Geschätzte Höhe: P99 des pixelweisen DOM1−DGM1 innerhalb eines Kronenkandidaten. Automatische Abgrenzung und Gebäudefilter können Fehler enthalten.
- Die Modelle stehen bereits auf basemap.de DGM5. Die Geländehöhe darf nicht noch einmal addiert werden. Die Modellform ist schematisch.
