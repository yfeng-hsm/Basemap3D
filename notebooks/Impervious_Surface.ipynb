# Imperviousness as a WMS layer
Two independent raster layers: aerial imagery and Copernicus Imperviousness Density 2024 (10 m; values 0–100%).


```python
from basemap3d import Map, WmsTileLayer

m = Map(location=[50.0, 8.27], distance_m=4000,
        buildings=False, heading=0, pitch=-70)
WmsTileLayer(opacity=0.45).add_to(m)

wms = "https://geoserver.geoville.com/geoserver/nvlcc_2024/wms"
WmsTileLayer(
    url=wms, layers="HRL_NVLCC_IMD_2024_10m",
    name="Imperviousness · 2024 · 10 m", opacity=0.8,
    description="Copernicus IMD · 2024 · 10 m · 0–100 %",
    attribution="© European Union · Copernicus Land Monitoring Service",
    legend_url=wms + "?service=WMS&request=GetLegendGraphic"
                     "&version=1.1.1&format=image/png"
                     "&layer=HRL_NVLCC_IMD_2024_10m",
).add_to(m)
m
```
