## 1 · Bäume als eigener 3D-Layer

Laden Sie 40.482 automatische Baumkandidaten aus einem 7 × 4 km großen Bearbeitungsrahmen mit 28 Rasterzellen. Die Darstellung verwendet geschätzte Baumhöhen. Schalten Sie den Layer aus und wieder ein; klicken Sie einen Baum an, um seine ID und Höhenangaben zu prüfen.


```python
from basemap3d import Map, TreeLayer, WmsTileLayer
m = Map(location=[49.99892, 8.27412], zoom_start=16,
        title="Mainz · Bäume als 3D-Layer")
WmsTileLayer(opacity=0.7).add_to(m)
TreeLayer().add_to(m)
m.save("results/mainz_trees.html")
m
```

## 2 · Geschätzte Höhe als Startansicht

Erstellen Sie eine unabhängige zweite Karte. Vergleichen Sie die räumliche Verteilung mit der einheitlichen Darstellung. Die Quelle enthält 109 Höhenwerte über 30 m, die geprüft werden sollen. Ein Kandidat ist kein bestätigter Baum.


```python
m_estimated = Map(location=[49.99892, 8.27412], zoom_start=16,
                  title="Mainz · Geschätzte Baumhöhe")
WmsTileLayer(opacity=0.7).add_to(m_estimated)
TreeLayer().add_to(m_estimated)
m_estimated.save("results/mainz_trees_estimated.html")
m_estimated
```

## 3 · Schatten von Bäumen und Gebäuden

Untersuchen Sie denselben Ausschnitt mit geschätzten Baumhöhen und Sonnenstand. Vergleichen Sie am 21. Juni 09:00, 13:00 und 17:00 Uhr (Europe/Berlin). Halten Sie die Kamera fest und schalten Sie Bäume bzw. Gebäude einzeln aus und ein. Wechseln Sie anschließend auf den 21. Dezember und vergleichen Sie Richtung und Länge der Schatten. Die Darstellung verwendet ausschließlich geschätzte Baumhöhen.

Das Luftbild lässt sich im Layer-Menü ausblenden oder in der Deckkraft anpassen. Bereits fotografierte Schatten bewegen sich beim Ändern der Uhrzeit nicht; blenden Sie das Luftbild für einen Vergleich der Modellschatten aus. Die Modellschatten sind eine qualitative Illustration. Schematische, lichtundurchlässige Kronen und unsichere Höhen erlauben keine belastbare Aussage zu Beschattungsfläche, Kühlung oder Aufenthaltskomfort. Blattverlust und Lichtdurchlässigkeit werden nicht simuliert.

**Ergebnis:** Drei Ansichten mit gleicher Kamera und unterschiedlichen Uhrzeiten. Beschreiben Sie, wo Bäume zusätzliche Schatten erzeugen und wo bereits Gebäudeschatten liegen.


```python
from basemap3d import SunShadows
m_shadows = Map(location=[49.99892, 8.27412], distance_m=450,
                title="Mainz · Baum- und Gebäudeschatten")
WmsTileLayer(opacity=0.7).add_to(m_shadows)
TreeLayer().add_to(m_shadows)
SunShadows(date="2026-06-21", hour=17).add_to(m_shadows)
m_shadows.save("results/mainz_trees_shadows.html")
m_shadows
```

## Einordnung und Transfer
