## 1 · Base map & 3D buildings
Start with a map. Terrain and LoD2 buildings load automatically.


```python
from basemap3d import Map

m = Map(location=[50.0063, 8.2692], zoom_start=16)
m
```

## 2 · Raster imagery via WMS
Choose a service URL and layer name: aerial imagery here, imperviousness in the extra example. Combine multiple WMS layers with independent controls.


```python
from basemap3d import Map

m = Map(location=[50.0063, 8.2692], zoom_start=16)
from basemap3d import WmsTileLayer

WmsTileLayer(
    url="https://geo4.service24.rlp.de/wms/rp_dop20.fcgi",
    layers="rp_dop20", name="Aerial imagery",
    opacity=0.7,
).add_to(m)
m
```

## 3 · Your polygons
Load GeoJSON polygons from a file, or replace them using the viewer upload. Click a feature to inspect its attributes.


```python
from basemap3d import Map

m = Map(location=[50.0063, 8.2692], zoom_start=16)
from basemap3d import WmsTileLayer, GeoJson

WmsTileLayer(opacity=0.7).add_to(m)
GeoJson("data/parking.geojson", popup=True).add_to(m)
m
```

## 4 · Query nearby buildings
Upload your own GeoJSON polygons. Buildings load for the new extent and are selected using the search distance.


```python
from basemap3d import (Map, WmsTileLayer, GeoJson,
                       NearbyBuildings)

m = Map(location=[50.0063, 8.2692], zoom_start=16)
WmsTileLayer(opacity=0.7).add_to(m)
GeoJson("data/parking.geojson", popup=True).add_to(m)
NearbyBuildings(distance_m=75).add_to(m)
m
```

## 5 · Trees & aerial imagery
Combine 3D trees with aerial imagery. This example has no polygons or building query.


```python
from basemap3d import Map, WmsTileLayer, TreeLayer

m = Map(location=[49.99892, 8.27412], distance_m=450)
WmsTileLayer(opacity=0.7).add_to(m)
TreeLayer().add_to(m)
m
```

## 6 · Trees, imagery & shadows
Add date and time controls to the tree scene. No polygons or building query.


```python
from basemap3d import Map, WmsTileLayer, TreeLayer

m = Map(location=[49.99892, 8.27412], distance_m=450)
WmsTileLayer(opacity=0.7).add_to(m)
TreeLayer().add_to(m)
from basemap3d import SunShadows

SunShadows(date="2026-06-21", hour=17).add_to(m)
m
```

## Complete · All visual modules
Combine aerial imagery, imperviousness (10 m), polygons, trees and shadows. Spatial queries are covered separately in example 4.


```python
from basemap3d import (Map, WmsTileLayer, GeoJson,
                       TreeLayer, SunShadows)

m = Map(location=[50.0063, 8.2692], zoom_start=16)
WmsTileLayer(opacity=0.7).add_to(m)

wms = "https://geoserver.geoville.com/geoserver/nvlcc_2024/wms"
WmsTileLayer(
    url=wms, layers="HRL_NVLCC_IMD_2024_10m",
    name="Imperviousness · 2024 · 10 m", opacity=0.8, show=False,
    description="Copernicus IMD · 2024 · 10 m · 0–100 %",
    attribution="© European Union · Copernicus Land Monitoring Service",
    legend_url=wms + "?service=WMS&request=GetLegendGraphic"
                     "&version=1.1.1&format=image/png"
                     "&layer=HRL_NVLCC_IMD_2024_10m",
).add_to(m)
GeoJson("data/parking.geojson", popup=True).add_to(m)
TreeLayer().add_to(m)
SunShadows(date="2026-06-21", hour=17).add_to(m)
m
```
