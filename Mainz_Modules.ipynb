## 1 · Basiskarte & 3D-Gebäude
Starten Sie mit einer Karte. Gelände und LoD2-Gebäude werden automatisch geladen.


```python
from basemap3d import Map

m = Map(location=[50.0063, 8.2692], zoom_start=16)
m
```

## 2 · Raster über WMS
URL und Layername bestimmen den Rasterinhalt: hier Luftbild, im Extra-Beispiel Bodenversiegelung. Mehrere WMS-Layer lassen sich unabhängig kombinieren.


```python
from basemap3d import Map

m = Map(location=[50.0063, 8.2692], zoom_start=16)
from basemap3d import WmsTileLayer

WmsTileLayer(
    url="https://geo4.service24.rlp.de/wms/rp_dop20.fcgi",
    layers="rp_dop20", name="Luftbild",
    opacity=0.7,
).add_to(m)
m
```

## 3 · Eigene Flächen
Laden Sie GeoJSON-Polygone aus einer Datei oder ersetzen Sie alle Flächen direkt über den Upload im Viewer. Ein Klick öffnet ihre Attribute.


```python
from basemap3d import Map

m = Map(location=[50.0063, 8.2692], zoom_start=16)
from basemap3d import WmsTileLayer, GeoJson

WmsTileLayer(opacity=0.7).add_to(m)
GeoJson("data/parking.geojson", popup=True).add_to(m)
m
```

## 4 · Gebäude räumlich abfragen
Ersetzen Sie die Flächen durch eine eigene GeoJSON-Datei. Die Gebäude werden für den neuen Ausschnitt geladen und anhand des Suchabstands neu ausgewählt.


```python
from basemap3d import (Map, WmsTileLayer, GeoJson,
                       NearbyBuildings)

m = Map(location=[50.0063, 8.2692], zoom_start=16)
WmsTileLayer(opacity=0.7).add_to(m)
GeoJson("data/parking.geojson", popup=True).add_to(m)
NearbyBuildings(distance_m=75).add_to(m)
m
```

## 5 · Bäume & Luftbild
Kombinieren Sie 3D-Bäume mit einem Luftbild. Dieses Beispiel enthält keine Flächen und keine Gebäudeabfrage.


```python
from basemap3d import Map, WmsTileLayer, TreeLayer

m = Map(location=[49.99892, 8.27412], distance_m=450)
WmsTileLayer(opacity=0.7).add_to(m)
TreeLayer().add_to(m)
m
```

## 6 · Bäume, Luftbild & Schatten
Ergänzen Sie Datum und Uhrzeit zum Baumbeispiel. Ohne Flächen und ohne Gebäudeabfrage.


```python
from basemap3d import Map, WmsTileLayer, TreeLayer

m = Map(location=[49.99892, 8.27412], distance_m=450)
WmsTileLayer(opacity=0.7).add_to(m)
TreeLayer().add_to(m)
from basemap3d import SunShadows

SunShadows(date="2026-06-21", hour=17).add_to(m)
m
```

## Komplett · Alle Module
Die vollständige Darstellung mit Luftbild, Bodenversiegelung (10 m), Flächen, Bäumen und Schatten. Die räumliche Abfrage wird separat in Beispiel 4 behandelt.


```python
from basemap3d import (Map, WmsTileLayer, GeoJson,
                       TreeLayer, SunShadows)

m = Map(location=[50.0063, 8.2692], zoom_start=16)
WmsTileLayer(opacity=0.7).add_to(m)

wms = "https://geoserver.geoville.com/geoserver/nvlcc_2024/wms"
WmsTileLayer(
    url=wms, layers="HRL_NVLCC_IMD_2024_10m",
    name="Bodenversiegelung · 2024 · 10 m", opacity=0.8, show=False,
    description="Copernicus IMD · 2024 · 10 m · 0–100 %",
    attribution="© European Union · Copernicus Land Monitoring Service",
    legend_url=wms + "?service=WMS&request=GetLegendGraphic"
                     "&version=1.1.1&format=image/png"
                     "&layer=HRL_NVLCC_IMD_2024_10m",
).add_to(m)
GeoJson("data/parking.geojson", popup=True).add_to(m)
TreeLayer().add_to(m)
SunShadows(date="2026-06-21", hour=17).add_to(m)
m
```
