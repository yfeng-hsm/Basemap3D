# Koblenz · Neues Gebäude und Schatten

Vorschau mit Basemap3D: LoD2 oder Stadtmesh, interaktive Gebäudeplatzierung und Sonnenstand.

**Ausführen:** Erst Installation, dann Karten-Zelle. Kein GPU-Runtime nötig.


```
import sys, subprocess, pathlib, os
repo = pathlib.Path('/content/koblenz-basemap3d')
if not repo.exists():
    subprocess.check_call(['git', 'clone', '--depth', '1', 'https://github.com/yfeng-hsm/GISProjekt26_HS_Mainz.git', str(repo)])
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', '-e', str(repo)])
sys.path.insert(0, str(repo))
os.chdir(repo)
from basemap3d import Map, Tiles3D, SunShadows, HousePlacement
```

```
#@title Koblenz: Modell auswählen und Zelle ausführen
Stadtmodell = "LoD2" #@param ["LoD2", "Mesh"]
Datum = "2026-06-21" #@param {type:"date"}
Uhrzeit = 16.0 #@param {type:"slider", min:4, max:22, step:0.25}

m = Map(location=[50.3604, 7.6036], distance_m=900,
        heading=25, pitch=-40, height=780,
        buildings=(Stadtmodell == "LoD2"),
        title="Koblenz · Gebäude und Schatten · " + Stadtmodell)
if Stadtmodell == "Mesh":
    Tiles3D(
        "https://dz.koblenz.de/assets/mesh_3D_2026/tileset.json",
        name="Koblenz · Stadtmesh 2026",
        attribution="Stadt Koblenz · GeoPortal Koblenz",
        popup=False,
    ).add_to(m)
SunShadows(date=Datum, hour=Uhrzeit).add_to(m)
HousePlacement().add_to(m)
m.show()
```

## Im Kartenfenster

1. Auf einen freien Platz zoomen.
2. Gebäudevorlage wählen → **Choose location** → auf freien Boden klicken.
3. Größe und Drehung einstellen, dann bestätigen.
4. Schatten aktivieren und Datum/Uhrzeit im Kartenfenster ändern.

**Modellwechsel:** Oben LoD2 oder Mesh wählen und Karten-Zelle erneut ausführen. Dabei werden platzierte Gebäude zurückgesetzt.

**Grenzen:** Gebäude sind schematisch; keine Kollisionsprüfung. Schatten in Fototexturen bleiben unverändert. Platzierte Gebäude werden in dieser Version weder in Python zurückgeschrieben noch im HTML-Export gespeichert.

Daten: basemap.de und Stadt Koblenz; Quellenhinweise im Viewer beachten.
